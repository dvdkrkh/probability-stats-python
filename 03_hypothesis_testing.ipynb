# Проверка статистических гипотез

**Темы:** нулевая и альтернативная гипотезы, уровень значимости, критическая область.

**Идея ноутбука:** с помощью симуляции посмотрим, какие результаты могли бы получиться при верной нулевой гипотезе, и сравним с ними результат наблюдаемой выборки.

## Подготовка

```python
import numpy as np
from scipy import stats

np.random.seed(42)
```

## 1. Проверка гипотезы по критической области

Сравним значение критерия, рассчитанное по наблюдаемой выборке, с критической границей.

### 1.1 Проверка гипотезы о средней высоте деревьев

Проверим, есть ли основания считать, что средняя высота деревьев в лесу больше 2,5 м. Зададим уровень значимости 5%. При $H_0$ примем, что высота деревьев распределена нормально со средним 2,5 м и стандартным отклонением 0,8 м.

Смоделируем 10 000 выборок по 20 деревьев при $H_0$ и по их средним найдём правую критическую границу. Затем сравним с ней среднее одной наблюдаемой выборки из 20 деревьев — 2,8 м. Запишем статистический вывод и объясним, что он означает для средней высоты деревьев в лесу.

```python
hyp_n = 10000
hyp_m = 20
hyp_alpha = 0.05 # уровень значимости
hyp_mu = 2.5
hyp_sigma = 0.8
hyp_table = np.random.normal(hyp_mu, hyp_sigma, (hyp_n, hyp_m))
hyp_mean = hyp_table.mean(axis=1) 
hyp_critlevel = np.quantile(hyp_mean, 1-hyp_alpha)# находит точку, ниже которой находятся примерно 95% чисел в переданном массиве
print(f"Критическая граница, выше которой находятся 5%: {hyp_critlevel:.3f}")
hyp_observed = 2.8 # наблидаемое среднее из 1 выборки 
if hyp_critlevel>=hyp_observed:
    print("На имеющихся данных на уровне значимости 5% нет оснований отвергнуть нулевую гипотезу в пользу альтернативы")
else:
    print("На имеющихся данных на уровне значимости 5% есть основания отвергнуть нулевую гипотезу в пользу альтернативы")
    
# сравним с теоретической границей
hyp_tail = hyp_alpha
hyp_area = 1 - hyp_tail
hyp_t = stats.norm.ppf(1 - hyp_alpha)
hyp_delta = hyp_t * (hyp_sigma / np.sqrt(hyp_m))
hyp_upper  = hyp_mu + hyp_delta
print(f"Теоретическая граница: {hyp_upper:.3f}")
```

**Вывод:**

При уровне значимости 5% критическая граница получилась примерно 2,784 м. Среднее наблюдаемой выборки — 2,8 м — оказалось выше этой границы и попало в критическую область.

Поэтому на имеющихся данных есть основания отвергнуть нулевую гипотезу о том, что средняя высота деревьев в лесу равна 2,5 м, в пользу альтернативной — средняя высота больше 2,5 м.
